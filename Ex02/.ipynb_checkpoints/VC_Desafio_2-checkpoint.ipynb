## Desafio 2 da Trilha de Visão Computacional — Solução Alternativa

Este notebook resolve o **mesmo desafio** do gabarito (reconhecimento facial, tipo controle de acesso a área restrita), mas com uma implementação um pouco diferente, para fins de estudo e comparação.

O pipeline geral é o mesmo do gabarito, porque ele vem da própria arquitetura clássica de reconhecimento facial com OpenCV:

1. **Detector de rosto** (`res10_300x300_ssd_iter_140000.caffemodel` + `deploy.prototxt`) — uma rede neural (SSD) que encontra a *posição* dos rostos numa imagem.
2. **Extrator de características / embedding** (`openface_nn4_small2_v1.t7`) — uma rede neural (OpenFace) que transforma cada rosto recortado num vetor numérico de 128 posições. Rostos da mesma pessoa geram vetores parecidos entre si; rostos de pessoas diferentes geram vetores distantes.
3. **Classificador** — decide, a partir do vetor, de quem é o rosto.

**A diferença principal desta versão está no passo 3.** Veja o resumo, com mais detalhes no final do notebook:

| | Gabarito | Esta versão |
|---|---|---|
| Classificador | SVM (`sklearn.svm.SVC`) treinado com todos os embeddings | Sem treinar modelo nenhum: calcula o **centróide** (vetor médio) dos embeddings de cada pessoa e compara por **similaridade de cosseno** |
| Reconhece "desconhecido"? | Não — sempre escolhe uma das pessoas cadastradas, mesmo com baixa confiança | Sim — se a similaridade não passar de um limiar, retorna `"Desconhecido"` |
| Código de detecção + embedding | Duplicado (uma vez na célula de treino, outra na célula de teste) | Escrito uma única vez em funções reutilizáveis |
| Teste | Uma imagem fixa, escolhida manualmente | Roda automaticamente em **todas as imagens da pasta `test/`** e mostra tudo em um grid |

### Estrutura de pastas esperada

```
dataset/

    pessoa_2/   -> fotos da pessoa 2
```

> **Observação sobre o dataset:** neste pacote, cada pasta de `dataset/` tem só 1 ou 2 fotos de exemplo (as que vieram junto com o desafio). O ideal, como no enunciado original, é ter várias fotos por pessoa (ex: 7) — quanto mais fotos, mais robusto fica o reconhecimento. Basta adicionar mais imagens dentro da pasta de cada pessoa; o código já é genérico e funciona com qualquer quantidade.

> **Sobre a pasta `test/`:** ela já vem com cópias das próprias fotos do `dataset/`, só para confirmar que o pipeline funciona de ponta a ponta (mas isso não é um teste justo, já que são as mesmas fotos usadas no treino!). Quando você tiver fotos novas para testar, é só **jogar os arquivos dentro da pasta `test/`** e rodar novamente a célula de teste — não precisa mudar nada no código.

````
import cv2
import matplotlib.pyplot as plt
import os
import glob
import pickle
import numpy as np
import imutils
from tqdm.notebook import tqdm
````

### Conferir a Estrutura de Pastas

Antes de processar qualquer coisa, vale a pena confirmar que o Jupyter está enxergando as pastas certas — esse é o erro mais comum nesse tipo de projeto (arquivo "não encontrado" por estar no diretório errado).

````
print("Pasta atual:", os.getcwd())
print()

for pessoa in sorted(os.listdir("dataset")):
    caminho = os.path.join("dataset", pessoa)
    if os.path.isdir(caminho):
        n_fotos = len([f for f in os.listdir(caminho) if not f.startswith(".")])
        print(f"dataset/{pessoa}: {n_fotos} foto(s)")

n_teste = len([f for f in os.listdir("test") if not f.startswith(".")])
print(f"\ntest/: {n_teste} foto(s)")

assert os.path.exists("face_detection_model/deploy.prototxt"), "deploy.prototxt não encontrado — confira a pasta face_detection_model/"
assert os.path.exists("face_detection_model/res10_300x300_ssd_iter_140000.caffemodel"), "caffemodel não encontrado — confira a pasta face_detection_model/"
assert os.path.exists("openface_nn4_small2_v1.t7"), "openface_nn4_small2_v1.t7 não encontrado na pasta raiz do notebook"
print("\nTodos os arquivos de modelo foram encontrados.")
````

### Carregar os Modelos

Diferente do gabarito (que carrega os modelos de novo, do zero, na célula de treino e de novo na célula de teste), aqui carregamos **uma única vez** e reaproveitamos os objetos `detector` e `embedder` no resto do notebook.

````
detector = cv2.dnn.readNetFromCaffe(
    "face_detection_model/deploy.prototxt",
    "face_detection_model/res10_300x300_ssd_iter_140000.caffemodel"
)

embedder = cv2.dnn.readNetFromTorch("openface_nn4_small2_v1.t7")

print("Modelos carregados com sucesso.")
````

### Funções Reutilizáveis de Detecção e Extração de Embedding

No gabarito, o bloco de "detectar rosto + extrair embedding" aparece **duplicado**: uma vez na célula que processa o dataset, e outra (quase idêntica) na célula que testa uma imagem nova. Isso é um problema de manutenção: se você mudar algo em um lugar (ex: o limiar de confiança), precisa lembrar de mudar no outro também.

Aqui, escrevemos essa lógica **uma única vez**, em duas funções:

- `detectar_rostos(imagem, confianca_minima, tamanho_minimo)`: roda o detector SSD e devolve uma lista com todos os rostos encontrados (caixa delimitadora + confiança + recorte da imagem do rosto).
- `extrair_embedding(rosto_recortado)`: passa o recorte do rosto pela rede OpenFace e devolve o vetor de 128 posições.

Essas duas funções são usadas tanto para montar o "banco de rostos" quanto para testar imagens novas.

````
def detectar_rostos(imagem, confianca_minima=0.5, tamanho_minimo=20):
    """Detecta todos os rostos em uma imagem usando o detector SSD (Caffe).

    Retorna uma lista de dicionários, um por rosto encontrado, com:
      - 'caixa': (startX, startY, endX, endY)
      - 'confianca': confiança da detecção (0 a 1)
      - 'rosto': recorte da imagem (array numpy) só com o rosto
    """
    (h, w) = imagem.shape[:2]

    blob = cv2.dnn.blobFromImage(
        cv2.resize(imagem, (300, 300)), 1.0, (300, 300),
        (104.0, 177.0, 123.0), swapRB=False, crop=False
    )
    detector.setInput(blob)
    deteccoes = detector.forward()

    resultados = []
    for i in range(deteccoes.shape[2]):
        confianca = deteccoes[0, 0, i, 2]
        if confianca < confianca_minima:
            continue

        caixa = deteccoes[0, 0, i, 3:7] * np.array([w, h, w, h])
        (startX, startY, endX, endY) = caixa.astype("int")

        # Garante que a caixa está dentro dos limites da imagem
        startX, startY = max(0, startX), max(0, startY)
        endX, endY = min(w, endX), min(h, endY)

        rosto = imagem[startY:endY, startX:endX]
        (altura_rosto, largura_rosto) = rosto.shape[:2]

        if largura_rosto < tamanho_minimo or altura_rosto < tamanho_minimo:
            continue

        resultados.append({
            "caixa": (startX, startY, endX, endY),
            "confianca": float(confianca),
            "rosto": rosto,
        })

    return resultados


def extrair_embedding(rosto_recortado):
    """Recebe o recorte de um rosto e devolve o vetor de embedding (128 posições)."""
    blob_rosto = cv2.dnn.blobFromImage(
        rosto_recortado, 1.0 / 255, (96, 96), (0, 0, 0), swapRB=True, crop=False
    )
    embedder.setInput(blob_rosto)
    vetor = embedder.forward().flatten()

    # Normaliza o vetor para comprimento 1 (necessário para usar similaridade de cosseno depois)
    norma = np.linalg.norm(vetor)
    if norma > 0:
        vetor = vetor / norma

    return vetor
````

### Processar a Base de Dados (Dataset)

Aqui está a segunda diferença em relação ao gabarito: em vez de guardar **cada embedding individual** em uma lista para depois treinar um SVM, calculamos direto o **centróide** (a média dos vetores) de cada pessoa. O centróide funciona como um "vetor representante" daquela pessoa.

Para cada foto do dataset:
1. Detectamos o rosto de maior confiança na imagem (assumindo 1 rosto por foto de treino, igual ao gabarito).
2. Extraímos o embedding desse rosto.
3. Guardamos o embedding na lista da pessoa correspondente (nome da subpasta em `dataset/`).

No final, calculamos a média dos embeddings de cada pessoa — esse é o centróide.

````
embeddings_por_pessoa = {}  # nome da pessoa -> lista de vetores de embedding

caminhos_dataset = glob.glob(os.path.join("dataset", "*", "*"))
caminhos_dataset = [p for p in caminhos_dataset if not os.path.basename(p).startswith(".")]

for caminho in tqdm(caminhos_dataset):
    nome_pessoa = os.path.basename(os.path.dirname(caminho))

    imagem = cv2.imread(caminho)
    if imagem is None:
        print(f"Aviso: não consegui ler {caminho}, pulando.")
        continue

    imagem = imutils.resize(imagem, width=600)

    rostos = detectar_rostos(imagem)
    if len(rostos) == 0:
        print(f"Aviso: nenhum rosto encontrado em {caminho}, pulando.")
        continue

    # Assumindo 1 rosto por foto de treino: pega o de maior confiança
    melhor_rosto = max(rostos, key=lambda r: r["confianca"])

    vetor = extrair_embedding(melhor_rosto["rosto"])

    embeddings_por_pessoa.setdefault(nome_pessoa, []).append(vetor)

for pessoa, vetores in embeddings_por_pessoa.items():
    print(f"{pessoa}: {len(vetores)} embedding(s) extraído(s)")
````

````
# Calcula o centróide (vetor médio, renormalizado) de cada pessoa
centroides = {}

for pessoa, vetores in embeddings_por_pessoa.items():
    media = np.mean(vetores, axis=0)
    norma = np.linalg.norm(media)
    centroides[pessoa] = media / norma if norma > 0 else media

os.makedirs("output", exist_ok=True)
with open("output/centroides.pickle", "wb") as f:
    pickle.dump(centroides, f)

print("Centróides calculados e salvos em output/centroides.pickle:")
for pessoa in centroides:
    print(" -", pessoa)
````

### Reconhecer um Rosto por Similaridade de Cosseno

No gabarito, o reconhecimento é feito com `recognizer.predict_proba(vec)` de um SVM já treinado. Aqui, comparamos o vetor do rosto testado com **cada centróide**, usando **similaridade de cosseno** (quanto mais próximo de 1, mais parecido; quanto mais próximo de -1 ou 0, menos parecido).

A pessoa escolhida é a de maior similaridade — **mas só se essa similaridade passar de um limiar mínimo** (`limiar`). Se nenhuma pessoa bater esse mínimo, o rosto é classificado como `"Desconhecido"`. Essa é a melhoria em relação ao gabarito: o SVM do gabarito *sempre* escolhe uma das pessoas cadastradas, mesmo que o rosto seja de um completo estranho.

````
def reconhecer_rosto(vetor, centroides, limiar=0.5):
    """Compara o vetor de um rosto com os centróides de cada pessoa.

    Retorna (nome, similaridade). Se a melhor similaridade for menor que o limiar,
    retorna ("Desconhecido", similaridade).
    """
    melhor_nome = None
    melhor_similaridade = -1.0

    for nome, centroide in centroides.items():
        similaridade = float(np.dot(vetor, centroide))  # vetores já normalizados -> dot = cosseno
        if similaridade > melhor_similaridade:
            melhor_similaridade = similaridade
            melhor_nome = nome

    if melhor_similaridade < limiar:
        return "Desconhecido", melhor_similaridade

    return melhor_nome, melhor_similaridade
````

### Testar em Todas as Imagens da Pasta `test/`

Diferente do gabarito, que testa uma única imagem escolhida manualmente (`test/modelo_2_1.jpeg`), esta célula roda o reconhecimento em **todas as imagens da pasta `test/`** automaticamente e mostra o resultado de todas em um grid.

Quando você tiver as 5 fotos novas para testar, basta colocá-las dentro da pasta `test/` (pode apagar as fotos de exemplo que vieram com o pacote, se quiser) e rodar esta célula de novo — nenhuma outra alteração é necessária.

````
LIMIAR_RECONHECIMENTO = 0.5  # ajuste conforme a quantidade/qualidade das fotos do dataset

caminhos_teste = sorted(glob.glob(os.path.join("test", "*")))
caminhos_teste = [p for p in caminhos_teste if not os.path.basename(p).startswith(".")]

resultados_teste = []  # vai guardar (nome_arquivo, imagem_anotada)

for caminho in caminhos_teste:
    imagem = cv2.imread(caminho)
    if imagem is None:
        print(f"Aviso: não consegui ler {caminho}, pulando.")
        continue

    imagem = imutils.resize(imagem, width=600)
    rostos = detectar_rostos(imagem)

    for rosto_info in rostos:
        (startX, startY, endX, endY) = rosto_info["caixa"]
        vetor = extrair_embedding(rosto_info["rosto"])
        nome, similaridade = reconhecer_rosto(vetor, centroides, limiar=LIMIAR_RECONHECIMENTO)

        cor = (0, 0, 255) if nome == "Desconhecido" else (0, 255, 0)
        texto = f"{nome}: {similaridade:.2f}"
        y_texto = startY - 10 if startY - 10 > 10 else startY + 10

        cv2.rectangle(imagem, (startX, startY), (endX, endY), cor, 2)
        cv2.putText(imagem, texto, (startX, y_texto),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, cor, 2)

    resultados_teste.append((os.path.basename(caminho), cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)))

# Mostra tudo em um grid
n = len(resultados_teste)
colunas = min(3, n) if n > 0 else 1
linhas = (n + colunas - 1) // colunas if n > 0 else 1

fig, axs = plt.subplots(linhas, colunas, figsize=(6 * colunas, 5 * linhas))
axs = np.array(axs).reshape(-1) if n > 1 else [axs]

for ax, (nome_arquivo, img) in zip(axs, resultados_teste):
    ax.imshow(img)
    ax.set_title(nome_arquivo)
    ax.axis("off")

for ax in axs[len(resultados_teste):]:
    ax.axis("off")

plt.tight_layout()
plt.show()
````

## Diferenças em Relação ao Gabarito — Resumo

| Aspecto | Gabarito | Esta solução |
|---|---|---|
| Classificador | SVM (`sklearn.svm.SVC`) treinado com todos os embeddings individuais | Centróide (média) dos embeddings por pessoa + similaridade de cosseno — sem "treinar" nada |
| Rosto desconhecido | Sempre atribui uma das pessoas cadastradas, mesmo com baixa confiança | Retorna `"Desconhecido"` se a similaridade ficar abaixo do limiar |
| Organização do código | Bloco de detecção + embedding duplicado (treino e teste) | Duas funções reutilizáveis (`detectar_rostos`, `extrair_embedding`) usadas nos dois lugares |
| Escala de teste | Uma imagem fixa, escolhida manualmente | Todas as imagens da pasta `test/`, automaticamente, mostradas em grid |
| Adicionar uma pessoa nova | Precisa reprocessar todo o dataset e re-treinar o SVM do zero | Só precisa calcular o centróide da pessoa nova — as outras pessoas não são afetadas |

**Limitações desta abordagem (para pensar / evoluir depois):**
- Com poucas fotos por pessoa (como neste pacote de exemplo), o centróide fica pouco robusto a variações de ângulo, iluminação e expressão. O ideal é ter várias fotos por pessoa, como no enunciado original (7 fotos).
- O valor do `LIMIAR_RECONHECIMENTO` (0.5) foi escolhido de forma arbitrária — na prática, esse valor deveria ser ajustado testando com fotos de pessoas conhecidas e desconhecidas, medindo a taxa de falsos positivos/negativos.
- Um classificador treinado (como o SVM do gabarito, ou um k-NN) pode aprender fronteiras de decisão mais sofisticadas do que "distância até a média" — a abordagem por centróide é mais simples e interpretável, mas menos poderosa quando o dataset é grande e variado.
