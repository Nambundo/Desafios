## Desafio 1 da Trilha de Visão Computacional — Solução Alternativa

Este notebook resolve o **mesmo desafio** (borrar o fundo de uma imagem contendo uma pessoa, tipo efeito de vídeo chamada), mas com uma **implementação diferente** da que aparece no gabarito oficial, para fins de estudo e comparação.

**Principais diferenças em relação ao gabarito:**

1. Em vez de recortar a face com fatiamento retangular (`image[y:h, x:w]`) e colar de volta, aqui construímos uma **máscara** (imagem preta e branca) que marca onde está a face, e usamos essa máscara para combinar a imagem original com a imagem borrada. Isso é o mesmo princípio usado em ferramentas de recorte/composição em Visão Computacional (ex: `cv2.bitwise_and`, blending com máscara).
2. A máscara usa uma **elipse** ao invés de um retângulo, e passa por um **blur** para suavizar as bordas — isso evita o "corte" abrupto e visível entre a área nítida do rosto e o fundo borrado que aparece na solução do gabarito.
3. O código funciona corretamente para **múltiplas faces** na imagem. No gabarito, se `detectMultiScale` encontrar mais de um rosto, o `for` sobrescreve as variáveis `x, y, width, height, face_img` a cada iteração e, no final, só a **última face detectada** fica nítida — as demais ficam borradas. Aqui, cada face detectada é adicionada à máscara.
4. No final, comparamos visualmente a imagem original, a máscara, a versão totalmente borrada e o resultado final lado a lado.

Ao final do notebook há uma seção explicando essas diferenças em mais detalhes.

### Instalação e Importação da Biblioteca

````
import cv2
import numpy as np
import matplotlib.pyplot as plt
````

### Ler Imagem

Lemos a imagem com `cv2.imread` (que carrega em BGR) e convertemos para RGB apenas para exibição com o `matplotlib` (o processamento em si pode ser feito em BGR ou RGB, já que só estamos aplicando blur e detecção — não faz diferença de canal aqui).

````
image_bgr = cv2.imread("desafio_1.jpeg")
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(image_rgb)
plt.title("Imagem original")
plt.axis("off")
plt.show()
````

### Transformar em Escala de Cinza

O `haarcascade` espera uma imagem em tons de cinza para detectar rostos.

````
image_gray = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(6, 6))
plt.imshow(image_gray, cmap="gray")
plt.title("Imagem em escala de cinza")
plt.axis("off")
plt.show()
````

### Detectar as Faces

Usamos o classificador Haar Cascade para encontrar as faces na imagem, assim como no gabarito. A diferença começa aqui: em vez de guardar apenas as coordenadas da **última** face do laço, guardamos uma **lista com todas as faces** detectadas (com padding aplicado), já pronta para ser usada na criação da máscara mais adiante.

````
detec = cv2.CascadeClassifier("haarcascade_frontalface_default.xml")
faces = detec.detectMultiScale(image_gray, scaleFactor=1.3, minNeighbors=3)

padding = 40  # pixels de folga ao redor do rosto detectado

altura_img, largura_img = image_rgb.shape[:2]
copy_image_rgb = image_rgb.copy()

faces_com_padding = []  # vai guardar (x1, y1, x2, y2) de cada rosto, já com padding

for (x, y, larg, alt) in faces:
    # Desenha o retângulo original (sem padding) só para visualização
    cv2.rectangle(copy_image_rgb, (x, y), (x + larg, y + alt), (0, 255, 0), 3)

    x1 = max(0, x - padding)
    y1 = max(0, y - padding)
    x2 = min(largura_img, x + larg + padding)
    y2 = min(altura_img, y + alt + padding)

    faces_com_padding.append((x1, y1, x2, y2))

print(f"{len(faces_com_padding)} rosto(s) detectado(s).")

plt.figure(figsize=(6, 6))
plt.imshow(copy_image_rgb)
plt.title("Faces detectadas (retângulo verde)")
plt.axis("off")
plt.show()
````

### Criar uma Máscara Suave para a Região do Rosto

Esta é a principal diferença em relação ao gabarito. Em vez de simplesmente colar o recorte retangular da face por cima da imagem borrada, criamos uma **máscara** em tons de cinza:

- Começamos com uma imagem toda preta (valor 0).
- Para cada rosto detectado, desenhamos uma **elipse branca** (valor 255) preenchida na região do rosto.
- Aplicamos um `GaussianBlur` na própria máscara, o que suaviza a transição entre 0 e 255 nas bordas.

Depois, essa máscara será usada como peso para misturar a imagem nítida (rosto) com a imagem borrada (fundo): quanto mais branco o pixel da máscara, mais peso a imagem original tem; quanto mais preto, mais peso a imagem borrada tem.

````
mask = np.zeros((altura_img, largura_img), dtype=np.uint8)

for (x1, y1, x2, y2) in faces_com_padding:
    centro = ((x1 + x2) // 2, (y1 + y2) // 2)
    eixos = ((x2 - x1) // 2, (y2 - y1) // 2)
    cv2.ellipse(mask, centro, eixos, angle=0, startAngle=0, endAngle=360,
                color=255, thickness=-1)  # thickness=-1 preenche a elipse

# Suaviza as bordas da máscara (evita corte abrupto entre nítido e borrado)
mask_suave = cv2.GaussianBlur(mask, (51, 51), 0)

plt.figure(figsize=(6, 6))
plt.imshow(mask_suave, cmap="gray")
plt.title("Máscara suave (branco = mantém nítido, preto = borra)")
plt.axis("off")
plt.show()
````

### Borrar a Imagem Inteira

Igual ao gabarito, aplicamos `GaussianBlur` na imagem inteira. Aqui usamos um kernel um pouco maior (`(31, 31)`) para deixar o efeito de fundo desfocado mais evidente.

````
blured_image = cv2.GaussianBlur(image_rgb, (31, 31), 20)

plt.figure(figsize=(6, 6))
plt.imshow(blured_image)
plt.title("Imagem totalmente borrada")
plt.axis("off")
plt.show()
````

### Compor a Imagem Final Misturando com a Máscara

No gabarito, a composição é feita por fatiamento direto:
```python
final_image = blured_image
final_image[y:height, x:width, :] = face_img
```
Aqui, em vez de substituir um retângulo inteiro, fazemos uma **mistura ponderada** pixel a pixel usando a máscara suave como peso (técnica de *alpha blending*):

```
final = imagem_original * alpha + imagem_borrada * (1 - alpha)
```

onde `alpha` é a máscara normalizada entre 0 e 1. Isso produz uma transição gradual entre a área nítida do rosto e o fundo borrado, em vez de uma borda reta e visível.

````
# Normaliza a máscara para valores entre 0 e 1, e replica para os 3 canais (RGB)
alpha = (mask_suave.astype(np.float32) / 255.0)
alpha_3canais = np.repeat(alpha[:, :, np.newaxis], 3, axis=2)

original_float = image_rgb.astype(np.float32)
blur_float = blured_image.astype(np.float32)

final_image = (original_float * alpha_3canais + blur_float * (1 - alpha_3canais))
final_image = np.clip(final_image, 0, 255).astype(np.uint8)

plt.figure(figsize=(6, 6))
plt.imshow(final_image)
plt.title("Resultado final: fundo borrado, rosto nítido")
plt.axis("off")
plt.show()
````

### Comparação Lado a Lado

````
fig, axs = plt.subplots(1, 4, figsize=(20, 5))

axs[0].imshow(image_rgb)
axs[0].set_title("Original")

axs[1].imshow(mask_suave, cmap="gray")
axs[1].set_title("Máscara suave")

axs[2].imshow(blured_image)
axs[2].set_title("Totalmente borrada")

axs[3].imshow(final_image)
axs[3].set_title("Resultado final")

for ax in axs:
    ax.axis("off")

plt.tight_layout()
plt.show()
````

## Diferenças em Relação ao Gabarito — Resumo

| Aspecto | Gabarito | Esta solução |
|---|---|---|
| Como marcar a região do rosto | Recorte retangular via fatiamento (`image[y:h, x:w]`) | Máscara em escala de cinza com elipse preenchida |
| Borda entre rosto e fundo | Corte abrupto (reto, "picotado") | Transição suave (a máscara passa por `GaussianBlur`) |
| Múltiplos rostos | Bug: só o último rosto do laço fica nítido, pois as variáveis são sobrescritas a cada iteração | Todos os rostos detectados entram na máscara e ficam nítidos |
| Composição final | Substituição direta de uma região retangular | *Alpha blending* pixel a pixel usando a máscara como peso (0 a 1) |
| Formato da área nítida | Retângulo | Elipse (mais parecido com o formato de um rosto) |

**Por que isso importa na prática?** A técnica de máscara + blending é a base de recursos "reais" de blur de fundo (Zoom, Google Meet, Teams etc.), que usam modelos de segmentação para gerar uma máscara detalhada da pessoa e depois misturam a imagem nítida com a borrada exatamente dessa forma — só que a máscara vem de uma rede neural de segmentação em vez de uma elipse geométrica simples. Entender essa mecânica de blending com máscara é o passo natural entre este desafio (com Haar Cascade + elipse) e uma versão mais avançada usando segmentação semântica.

**Possíveis melhorias futuras:**
- Usar um modelo de segmentação de pessoa (ex: MediaPipe Selfie Segmentation) em vez de detecção de rosto, para borrar tudo, exceto a pessoa inteira (não só o rosto).
- Ajustar dinamicamente o tamanho do kernel de blur da máscara conforme o tamanho do rosto detectado.
- Tratar o caso de nenhum rosto detectado (hoje, se `faces` vier vazio, a máscara fica toda preta e a imagem final sai toda borrada).
