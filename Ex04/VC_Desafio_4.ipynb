```python
import ultralytics
import matplotlib.pyplot as plt
import torch
```

### Processar Dados



```python
# Baixa o modelo de pose da YOLO.
model = ultralytics.YOLO("yolo11n-pose.pt")

# Usa GPU (cuda) se disponível, caso contrário usa CPU.
device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)

# Processa a imagem.
# TODO testar outras imagens.
results = model("desafio_4.jpeg")  # predict on an image
```

### Visualizar Resultados

```python
from PIL import Image
import cv2

print(results[0])

# Gera visualização do resultado
im_array = results[0].plot()

# results[0].plot() retorna a imagem em BGR (padrão OpenCV),
# então convertemos para RGB antes de exibir com matplotlib.
rgb_img = cv2.cvtColor(im_array, cv2.COLOR_BGR2RGB)

# Visualize a imagem rgb_img
plt.figure(figsize=(10, 10))
plt.imshow(rgb_img)
plt.axis('off')
plt.show()
```
